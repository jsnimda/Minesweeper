# Procedural wallpaper preview

Each grid uses reproducible seeds. Change `DESKTOP_SIZE` to preview another resolution.


```python
import random

from IPython.display import display
from PIL import Image, ImageDraw

from synthetic_scenes.wallpaper import (
    render_procedural_wallpaper,
    sample_procedural_subtype,
)

DESKTOP_SIZE = (1280, 720)
THUMBNAIL_SIZE = (240, 135)
SEED = 1200


def show_grid(images, labels=None):
    columns = 5
    label_height = 20 if labels else 0
    rows = (len(images) + columns - 1) // columns
    grid = Image.new(
        "RGB",
        (columns * THUMBNAIL_SIZE[0], rows * (THUMBNAIL_SIZE[1] + label_height)),
        "white",
    )
    draw = ImageDraw.Draw(grid)
    for index, image in enumerate(images):
        x = (index % columns) * THUMBNAIL_SIZE[0]
        y = (index // columns) * (THUMBNAIL_SIZE[1] + label_height)
        image.thumbnail(THUMBNAIL_SIZE)
        grid.paste(image, (x, y))
        if labels:
            draw.text((x + 4, y + THUMBNAIL_SIZE[1] + 2), labels[index], fill="black")
    display(grid)
```

## solid


```python
images = [
    render_procedural_wallpaper(DESKTOP_SIZE, random.Random(SEED + 0 * 100 + index), "solid")
    for index in range(20)
]
show_grid(images)
```

## linear


```python
images = [
    render_procedural_wallpaper(DESKTOP_SIZE, random.Random(SEED + 1 * 100 + index), "linear")
    for index in range(20)
]
show_grid(images)
```

## radial


```python
images = [
    render_procedural_wallpaper(DESKTOP_SIZE, random.Random(SEED + 2 * 100 + index), "radial")
    for index in range(20)
]
show_grid(images)
```

## multi_gradient


```python
images = [
    render_procedural_wallpaper(DESKTOP_SIZE, random.Random(SEED + 3 * 100 + index), "multi_gradient")
    for index in range(20)
]
show_grid(images)
```

## Mixed procedural wallpapers

Subtypes are sampled using the production default weights.


```python
images = []
labels = []
for index in range(20):
    rng = random.Random(SEED + 1000 + index)
    subtype = sample_procedural_subtype(rng)
    images.append(render_procedural_wallpaper(DESKTOP_SIZE, rng, subtype))
    labels.append(subtype)
show_grid(images, labels)
```
